##### Locate Elements in the DOM
Can search by:
* ID
* CSS selector
* XPath
* Link text, tag name, or class name

using By class with .find_element() or .find_elements()

```python
from selenium import webdriver
from selenium.webdriver.firefox.options import Options
from selenium.webdriver.common.by import By

options = Options()
options.add_argument("--headless")
driver = webdriver.Firefox(options=options)
driver.implicitly_wait(5)

driver.get("https://bandcamp.com/discover/")
print(driver.title)

pagination_button = driver.find_element(By.ID, "view-more") # finding a single element
print(pagination_button.accessible_name)

tracks = driver.find_elements(By.CLASS_NAME,"results-grid-item") #finding all track elements currently visible
print(len(tracks))
print(tracks[0].text)
```

```python
# performing sub-searchs on elements
track_1 = tracks[0]
album = track_1.find_element(By.CSS_SELECTOR, "div.title a span")
print(album.text)
```

##### Send Keystrokes and Text Entry

```python
from selenium.webdriver.common.keys import Keys

options = Options()
options.add_argument("--headless")
driver = webdriver.Firefox(options=options)
driver.implicitly_wait(5)

driver.get("https://bandcamp.com/discover/")

search_box = driver.find_element(By.TAG_NAME, "input")
search_box.send_keys("Search for this")
search_box.submit()

# OR search_box.send_keys(Keys.ENTER)

time.sleep(0.5)

driver.quit()
```

##### Use Hover, Drag-and-Drop, and More Complex Gestures
using ActionChains class for actions like dragging, hovering, or right clicking

```python
from selenium.webdriver import ActionChains

driver = webdriver.Firefox()  # Not in headless mode
driver.implicitly_wait(5)

driver.get("https://bandcamp.com/discover/")

# Accept cookies, if required
try:
    cookie_accept_button = driver.find_element(
        By.CSS_SELECTOR,
        "#cookie-control-dialog button.g-button.outline",
    )
    cookie_accept_button.click()
except NoSuchElementException:
    pass

time.sleep(0.5)

menu = driver.find_element(By.CSS_SELECTOR, ".menu")
submenu = driver.find_element(By.CSS_SELECTOR, ".menu #submenu")

actions = ActionChains(driver) # instantiating ActionChains
actions.move_to_element(menu) # hover on the menu element, triggering dropdown
actions.click(submenu) # clicking on child element in submenu

actions.perform()
```

##### Understand Implicit, Explicit, and Fluent Waits
1. Implicit Wait: Selenium polls the DOM for a specified time whenever you try to find an element.
2. Explicit Wait: Selenium applies conditions to waits, which makes it more flexible.
3. Fluent Wait: Selenium allows you to specify the polling interval, ignore certain exceptions that occur during polling, and set custom timeout messages.


##### Set up an Implicit Wait
driver.implicitly_wait(5)

##### Use Explicit Waits for Targeted Waiting
* Use WebDriverWait object with predefined conditions

```python
import time
from selenium import webdriver
from selenium.common.exceptions import NoSuchElementException
from selenium.webdriver.firefox.options import Options
from selenium.webdriver.common.by import By

from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

options = Options()
options.add_argument("--headless")
driver = webdriver.Firefox(options=options)
driver.implicitly_wait(5)

driver.get("https://bandcamp.com/discover/")

tracks = driver.find_elements(By.CLASS_NAME, "results-grid-item")
print(len(tracks))

try:
    cookie_accept_button = driver.find_element(
        By.CSS_SELECTOR,
        "#cookie-control-dialog button.g-button.outline",
    )
    cookie_accept_button.click()
except NoSuchElementException:
    pass

#pagination_button = driver.find_element(By.ID, "view-more")
#pagination_button.click()

wait = WebDriverWait(driver,10)
wait.until(EC.element_to_be_clickable((By.ID, "view-more")))

tracks = driver.find_elements(By.CLASS_NAME, "results-grid-item")
print(len(tracks))

driver.quit()
```
